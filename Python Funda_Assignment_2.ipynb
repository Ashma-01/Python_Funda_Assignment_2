```python
age_list = [24, 25, 26, 27, 28]
name_list = ["Anu", "Balan", "Chithra", "Divya", "Manu"]
print(age_list)
print(name_list)
```

```python
name_list.append("Yazhini")
print(name_list)
```

```python
age_list.insert(2, 30)
print(age_list)
```

```python
name_list.remove("Yazhini")
print(name_list)
```

```python
age_list.pop()
```

```python
age_list.extend([29,30,26])
print(age_list)
```

```python
age_list.sort(reverse=True)
print(age_list)
```

```python
max_age=max(age_list)
min_age=min(age_list)
sum_age=sum(age_list)

print("Max age:",max_age)
print("Min age:",min_age)
print("Sum of ages:",sum_age)
```

```python
print("First element:",name_list[0])
print("Last element:",name_list[-1])
print("Index 2 to 4:",name_list[2:5])
print("Reversed name list:",name_list[::-1])
```

```python
student_marks={"Arun":85,"Bhanu":92,"kiran":78,"Deepak":65,"Maya":88}
print("Bhanu's mark:",student_marks["Bhanu"])
student_marks["Janani"]=80
student_marks["Arun"]=82

print("Keys:",list(student_marks.keys()))
print("Values:",list(student_marks.values()))
print("key-value pairs:",list(student_marks.items()))
```

```python
my_set= set(['a','e','i','o','u','a','a','i'])
print(my_set)

# sets in python only store unique elements and do not preserve order. The duplicate entries 'a' and 'i' are automatically removed,leaving{'a','e','i','o','u'}.
```

```python
my_set[4]='s'

# This line rises a TypeError:'set' object does not support item assignment.Sets are unordered collections and do not support indexing or item mutation via index locations.
```

```python
set1={1,3,5,7,9}
set2={2,3,5,8,10}

union_set= set1.union(set2)
intersection_set= set1.intersection(set2)

print("Union:",union_set)
print("inersection:",intersection_set)
```

```python
score= float(input("Enter your score(0 to 10): "))

if 0 <= score <=10:
    if score > 7:
        print("Above Average:Outstanding work! Keep up the excellent performance.")
    elif 4 <= score <=7:
        print("Average: Good effort! Keep practicing, there's room for improvement.")
    else:
        print("Below Average: Need to improve your performance, consistent practice will lead to better results.")
else:
    print("Invalid input! Please enter a score between 0 and 10.")        
```
